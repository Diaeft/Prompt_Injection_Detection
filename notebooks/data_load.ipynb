```python
from datasets import load_dataset
import pandas as pd
import os
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

hf_token = os.getenv("HF_TOKEN")

if hf_token:
    print("Connexion Hugging Face sécurisée et configurée !")
else:
    print("Erreur : Le token n'a pas été trouvé dans le fichier .env")
```

```python
ds = load_dataset("cyberec/Prompt-injection-dataset", "core")
print(ds)
```

```python
df_train = ds["train"].to_pandas()
df_val   = ds["validation"].to_pandas()
df_test  = ds["test"].to_pandas()

print("Train :",      len(df_train))   # 4391
print("Validation :", len(df_val))     # 941
print("Test :",       len(df_test))    # 942
```

```python
# Vérifier les colonnes et types
df_train.info()
```

```python
# Vérifier les valeurs manquantes
df_train.isnull().sum()
```

```python
# Voir la répartition des labels
print(df_train["label"].value_counts())
# 0 = bénin   (prompt normal)
# 1 = malicieux (prompt injection)

print(df_train["category"].value_counts())
print(df_train["severity"].value_counts())
```

```python
# Créer le dossier data/ s'il n'existe pas
os.makedirs("../data", exist_ok=True)

# Sauvegarder les 3 splits
df_train.to_csv("../data/train.csv", index=False)
df_val.to_csv("../data/val.csv",     index=False)
df_test.to_csv("../data/test.csv",   index=False)

print("✅ CSV sauvegardés dans data/")
```
